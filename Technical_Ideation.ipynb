# Technical Ideation: Predictive Maintenance LSTM Pipeline
This notebook handles data ingestion, temporal feature processing, sequence generation, and multi-output LSTM model assembly & training.

### 1. Load Libraries

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from keras.models import Model
from keras.layers import Input, LSTM, Dropout, Dense, Activation
import os
```

### 2. Ingest and Preprocess Raw Datasets

```python
# Load raw CSV files
machines = pd.read_csv('data/machines.csv')
errors = pd.read_csv('data/errors.csv')
failures = pd.read_csv('data/failures.csv')
data = pd.read_csv('data/telemetry.csv')
maintenances = pd.read_csv('data/maint.csv')

# One-hot encode categoricals
failures = failures.join(pd.get_dummies(failures['failure'])).drop('failure', axis=1)
errors = errors.join(pd.get_dummies(errors['errorID'])).drop('errorID', axis=1)

maintenances.loc[maintenances.comp == "comp1", "comp"] = "maintComp1"
maintenances.loc[maintenances.comp == "comp2", "comp"] = "maintComp2"
maintenances.loc[maintenances.comp == "comp3", "comp"] = "maintComp3"
maintenances.loc[maintenances.comp == "comp4", "comp"] = "maintComp4"
maintenances = maintenances.join(pd.get_dummies(maintenances['comp'])).drop('comp', axis=1)

# Aggregate events by Machine ID & Datetime
errors = errors.groupby(['machineID', 'datetime']).max().reset_index()
failures = failures.groupby(['machineID', 'datetime']).max().reset_index()
maintenances = maintenances.groupby(['machineID', 'datetime']).max().reset_index()

# Convert datetime format
failures['datetime'] = pd.to_datetime(failures['datetime'])
errors['datetime'] = pd.to_datetime(errors['datetime'])
data['datetime'] = pd.to_datetime(data['datetime'])
maintenances['datetime'] = pd.to_datetime(maintenances['datetime'])
```

### 3. Merge Datasets & Engineer Target Window (24h)

```python
# Merge operational telemetry data
machines = machines.drop('model', axis=1)
df = pd.merge(data, machines, on=['machineID'])
df = pd.merge(df, errors, how='left', on=['datetime', 'machineID']).fillna(0)
df = pd.merge(df, failures, how='left', on=['datetime', 'machineID']).fillna(0)

maintenances = maintenances[maintenances['datetime'] >= '2015']
df = pd.merge(df, maintenances, how='left', on=['datetime', 'machineID']).fillna(0)

# Engineer 24-hour target lookback failure window
day = 24
for comp_id in ['comp1', 'comp2', 'comp3', 'comp4']:
    col_target = f'{comp_id}_fails_in_aday'
    df[col_target] = df[comp_id]
    for i in range(len(df)):
        if df.iloc[i][comp_id] == 1:
            m_id = df.iloc[i]['machineID']
            for j in range(i, max(-1, i - day), -1):
                if df.iloc[j]['machineID'] != m_id:
                    break
                df.at[j, col_target] = 1
```

### 4. Normalization and Time-Series Sequence Generation

```python
features_col_name = ['volt', 'rotate', 'pressure', 'vibration', 'age', 'error1', 'error2', 'error3', 'error4', 'maintComp1', 'maintComp2', 'maintComp3', 'maintComp4']
target_col_day = ['comp1_fails_in_aday', 'comp2_fails_in_aday', 'comp3_fails_in_aday', 'comp4_fails_in_aday']

# Train/Test split across machines
train = pd.DataFrame(columns=df.columns)
test = pd.DataFrame(columns=df.columns)

for i in range(1, 101):
    train = train.append(df[df['machineID'] == i][:6503])
    test = test.append(df[df['machineID'] == i][6503:])

sc = MinMaxScaler()
train[features_col_name] = sc.fit_transform(train[features_col_name])
test[features_col_name] = sc.transform(test[features_col_name])

def gen_sequence(id_df, seq_length, seq_cols):
    df_zeros = pd.DataFrame(np.zeros((seq_length - 1, id_df.shape[1])), columns=id_df.columns)
    id_df = df_zeros.append(id_df, ignore_index=True)
    data_array = id_df[seq_cols].values
    num_elements = data_array.shape[0]
    lstm_array = []
    for start, stop in zip(range(0, num_elements - seq_length), range(seq_length, num_elements)):
        lstm_array.append(data_array[start:stop, :])
    return np.array(lstm_array)

def gen_label(id_df, seq_length, seq_cols, labels):
    df_zeros = pd.DataFrame(np.zeros((seq_length - 1, id_df.shape[1])), columns=id_df.columns)
    id_df = df_zeros.append(id_df, ignore_index=True)
    y_label = []
    num_elements = id_df.shape[0]
    for start, stop in zip(range(0, num_elements - seq_length), range(seq_length, num_elements)):
        y_label.append([id_df[labels[0]][stop], id_df[labels[1]][stop], id_df[labels[2]][stop], id_df[labels[3]][stop]])
    return np.array(y_label)

X_train = np.concatenate([gen_sequence(train[train['machineID'] == id_val], day, features_col_name) for id_val in train['machineID'].unique()])
y_train = np.concatenate([gen_label(train[train['machineID'] == id_val], day, features_col_name, target_col_day) for id_val in train['machineID'].unique()])

X_test = np.concatenate([gen_sequence(test[test['machineID'] == id_val], day, features_col_name) for id_val in test['machineID'].unique()])
y_test = np.concatenate([gen_label(test[test['machineID'] == id_val], day, features_col_name, target_col_day) for id_val in test['machineID'].unique()])

y_train = list(np.transpose(y_train))
y_test = list(np.transpose(y_test))
```

### 5. Construct & Train Multi-Output LSTM Network

```python
class UtkMultiOutputModel():
    def make_default_hidden_layers(self, inputs):
        x = LSTM(units=200, return_sequences=True)(inputs)
        x = Dropout(0.2)(x)
        x = LSTM(units=200, return_sequences=False)(inputs)
        x = Dropout(0.2)(x)
        return x

    def build_branch(self, inputs, name):
        x = self.make_default_hidden_layers(inputs)
        x = Dense(1)(x)
        return Activation("sigmoid", name=name)(x)

    def assemble_full_model(self, timestamp, nb_features):
        inputs = Input(shape=(timestamp, nb_features))
        b1 = self.build_branch(inputs, "comp1")
        b2 = self.build_branch(inputs, "comp2")
        b3 = self.build_branch(inputs, "comp3")
        b4 = self.build_branch(inputs, "comp4")
        return Model(inputs=inputs, outputs=[b1, b2, b3, b4], name="failure")

model = UtkMultiOutputModel().assemble_full_model(day, X_train.shape[2])

model.compile(
    optimizer='adam',
    loss={'comp1': 'binary_crossentropy', 'comp2': 'binary_crossentropy', 'comp3': 'binary_crossentropy', 'comp4': 'binary_crossentropy'},
    loss_weights={'comp1': 0.1, 'comp2': 0.1, 'comp3': 0.1, 'comp4': 0.1},
    metrics={'comp1': 'accuracy', 'comp2': 'accuracy', 'comp3': 'accuracy', 'comp4': 'accuracy'}
)

model.fit(X_train, y_train, epochs=1, batch_size=100, validation_split=0.2)

# Save trained model artifacts
os.makedirs('data/model', exist_ok=True)
with open("data/model/predictive_maint.json", "w") as json_file:
    json_file.write(model.to_json())
model.save_weights("data/model/predictive_maint.h5")
print("Model saved successfully to data/model/")
```
